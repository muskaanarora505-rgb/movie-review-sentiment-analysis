```python
import re
```

```python
def clean_text(text):
    # Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)

    # Convert to lowercase
    text = text.lower()

    # Remove special characters and numbers
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df['clean_review'] = df['review'].apply(clean_text)
```

```python
print("Original Review:")
print(df['review'].iloc[1])

print("\nCleaned Review:")
print(df['clean_review'].iloc[1])
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

print(list(stop_words)[:20])
```

```python
important_words = {'not', 'no', 'nor', 'never'}

stop_words = stop_words - important_words
```

```python
def remove_stopwords(tokens):
    return [
        word for word in tokens
        if word not in stop_words
    ]
```

```python
df['tokens_no_stopwords'] = df['tokens'].apply(remove_stopwords)
```

```python
print("Before:")
print(df['tokens'].iloc[0])

print("\nAfter stopword removal:")
print(df['tokens_no_stopwords'].iloc[0])
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
```

```python
def apply_stemming(tokens):
    return [stemmer.stem(word) for word in tokens]
```

```python
df['stemmed'] = df['tokens_no_stopwords'].apply(apply_stemming)
```

```python
print("Before stemming:")
print(df['tokens_no_stopwords'].iloc[0])

print("\nAfter stemming:")
print(df['stemmed'].iloc[0])
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def apply_lemmatization(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

df['lemmatized'] = df['tokens_no_stopwords'].apply(
    apply_lemmatization)

df['lemmatized_text'] = df['lemmatized'].apply(
    lambda x: ' '.join(x))
```
